## 4. Verificação da qualidade dos dados

Nesta etapa, serão investigados valores ausentes, registros duplicados e possíveis inconsistências nos conjuntos de dados. Nenhuma alteração será realizada antes de compreender o significado desses problemas.

### 4.1 Qualidade dos dados de jogadores

```python
df_players.duplicated().sum()
```

```python
missing_values_players = df_players.isna().sum()

missing_values_players[
    missing_values_players > 0
]
```

```python
missing_values_players_percentage = missing_values_players / df_players.shape[0] * 100

missing_values_players_percentage[
    missing_values_players_percentage > 0
]
```

#### Observações sobre a qualidade dos dados

- O conjunto não possui linhas completamente duplicadas.
- As colunas `damage`, `damage/min`, `kill_participation`, `kill_share` e `gold_share` possuem valores ausentes.
- As colunas com maior quantidade de valores ausentes são `damage` e `damage/min`, ambas com aproximadamente 68% dos seus valores ausentes.
- Inicialmente, os valores ausentes parecem estar concentrados em métricas que não eram registradas nas primeiras temporadas.

#### Investigação dos valores ausentes por temporada

A inspeção inicial sugere que os valores ausentes podem estar concentrados nas primeiras temporadas, quando algumas métricas ainda não eram registradas.

Para investigar essa hipótese, será comparada, em cada temporada, a quantidade total de registros com a quantidade de valores preenchidos nas colunas `damage` e `kill_participation`.

```python
players_by_season = df_players.groupby("season")["player"].count()

players_by_season
```

A seguir, será contabilizada a quantidade de valores não ausentes em `damage` para cada temporada.

```python
damage_values_by_season = df_players.groupby("season")["damage"].count()

damage_values_by_season
```

```python
kill_participation_by_season = df_players.groupby("season")["kill_participation"].count()

kill_participation_by_season
```

#### Conclusão da investigação

- A métrica `damage` só possui valores preenchidos a partir da temporada 10. 
- Nas temporadas em que `damage` é registrado, não existem valores ausentes nessa coluna.
- A métrica `kill_participation` só possui valores preenchidos a partid da temporada 4.
- Assim como ocorre em `damage`, a métrica `kill_participation` não possui valores ausentes nas temporadas em que essa métrica é registrada.  

Portanto, as ausências observadas não parecem ocorrer de maneira aleatória. Elas realmente estão associadas às temporadas em que essas métricas não eram registradas no conjunto de dados.

#### Verificação das demais métricas ausentes

Após identificar o padrão temporal de `damage` e `kill_participation`, será verificado se as demais colunas com valores ausentes seguem o mesmo comportamento.

```python
missing_metrics_by_season = df_players.groupby("season")[
    [
        "damage",
        "damage/min",
        "kill_participation",
        "kill_share",
        "gold_share"
    ]
].count()

missing_metrics_by_season
```

- As métricas `damage` e `damage/min` apresentam o mesmo padrão de preenchimento e só são registradas a partir da temporada 10.
- As métricas `kill_participation`, `kill_share` e `gold_share` apresentam o mesmo padrão e começam a ser registradas na temporada 4.
- Não existem ausências pontuais nas temporadas em que essas métricas são disponibilizadas.

#### Síntese da qualidade dos dados de jogadores

O conjunto não possui registros completamente duplicados. Os valores ausentes seguem um padrão temporal associado à disponibilidade das métricas em diferentes temporadas, não havendo sinais, até o momento, de ausências pontuais nas temporadas em que cada métrica é registrada.

### 4.2 Qualidade dos dados de campeões

```python
missing_values_champions_percentage = missing_values_champions / df_champions.shape[0] * 100

missing_values_champions_percentage[
    missing_values_champions_percentage > 0
]
```

#### Observações sobre a a qualidade dos dados

- O conjunto não possui linhas completamente duplicadas.

- Parte dos valores ausentes parece estar relacionada a métricas que ainda não eram registradas nas primeiras temporadas.
- Também é possível observar que, quando um campeão não possui partidas disputadas em determinada temporada e evento, suas estatísticas de desempenho aparecem como valores ausentes.

#### Investigação de valores ausentes causados por campeões não utilizados

A inspeção inicial sugere que os valores ausentes podem ter causas diferentes dentro do conjunto de dados de campeões.

Para investigar essa hipótese, serão analisadas as linhas em que a coluna `played_games` é igual a zero, a fim de verificar o impacto disso nas outras colunas.

```python
champions_without_games = df_champions[
    df_champions['played_games'] == 0
]

champions_without_games.head()
```

```python
champions_without_games.shape
```

```python
champions_without_games.isna().sum()
```

##### Conclusão sobre os campeões sem partidas

- 52 registros possuem `played_games` igual a zero.
- Em todos esses registros as seguintes métricas ficam ausentes: `played_by_number_of_players`, `win_rate`, `kills`, `deaths`, `assists`, `kill_death_assist_ratio`, `creep_score`, `cs/min`, `gold`, `gold/min`, `damage`, `damage/min`, `kill_participation`, `kill_share` e `gold_share`.
- Mesmo quando o campeão não é utilizado as seguintes métricas permanecem preenchidas: `season`, `event`, `champion`, `games_contests`, `pick_ban_ratio`, `banned_games`, `win` e `lose`
- Essa investigação explica parte das ausências do dataset, mas não explica completamente. Nos 52 registros em que `played_games` é igual a zero, `banned_games` não possui valores ausentes. Porém, analisando o conjunto como um todo, 28% dos valores de `banned_games` estão ausentes. 
- Outra parte dos valores ausentes está relacionada às temporadas iniciais, em que algumas métricas como `damage` e `damage/min` ainda não eram registradas.

#### Investigação dos valores ausentes em `banned_games`

A coluna `banned_games` possui aproximadamente 28% dos valores ausentes. Será investigado se essas ausências representam campeões que não foram banidos em nenhuma partida ou se indicam falta de informação. 

```python
champions_without_ban_data = df_champions[
    df_champions["banned_games"].isna()
]

champions_without_ban_data.head()
```

```python
champions_without_ban_data[
    [
        "season",
        "event",
        "champion",
        "games_contests",
        "banned_games",
        "played_games"
    ]
].head(20)
```

```python
(champions_without_ban_data["games_contests"] == champions_without_ban_data["played_games"]).value_counts()
```

##### Conclusão sobre valores ausentes em `banned_games`

- Em todos os registros em que `banned_games` está ausente, o valor de `games_contests` é igual ao de `played_games`. Dessa forma, nesses casos, todas as aparições do campeão no draft correspondem a partidas jogadas, não havendo registros de banimento.
- Sendo assim, há evidências consistentes de que os valores ausentes em `banned_games` representam zero banimentos e podem ser substituídos por 0 para fins de análise.

#### Investigação das métricas ausentes por temporada

Tendo identificado as ausências relacionadas a campeões sem partidas e os casos em que `banned_games` representa zero banimentos, será investigada a disponibilidade histórica das outras métricas.

Para garantir que campeões sem partidas não interfiram nessa análise, serão considerados somente os registros em que `played_games` é maior que zero.

```python
champions_with_games = df_champions[
    df_champions["played_games"] > 0
]

champions_with_games.shape
```

```python
champions_with_games_by_season = champions_with_games.groupby("season")["champion"].count()

champions_with_games_by_season
```

```python
champion_metrics_by_season = champions_with_games.groupby("season")[
    [
        "damage",
        "damage/min",
        "kill_participation",
        "kill_share",
        "gold_share"
    ]
].count()

champion_metrics_by_season
```

##### Conclusão sobre a disponibilidade histórica das métricas

- As métricas `damage` e `damage/min` começam a ser registradas a partir da temporada 10.
- As métricas `kill_participation`, `kill_share` e `gold_share` começam a ser registradas a partir da temporada 4.
- Existe preenchimento parcial das métricas `kill_participation`, `kill_share` e `gold_share` nas temporadas 7 e 8. Logo, a dispobinilidade dessas métricas nessas temporadas precisa ser investigada com mais detalhes.

#### Investigação do preenchimento parcial nas temporadas 7 e 8

Apesar de `kill_participation`, `kill_share` e `gold_share` começarem a ser registradas na temporada 4, foi identificado preenchimento parcial nas temporadas 7 e 8.

Será investigado se essas ausências estão relacionadas às diferentes etapas do campeonato representadas pela coluna `event`.

```python
champions_season_7_and_8 = champions_with_games[
    champions_with_games['season'].isin([7,8])
]

champions_season_7_and_8.shape
```

```python
champions_season_7_and_8.groupby(
    ["season", "event"]
    )["champion"].count()
```

```python
champions_season_7_and_8.groupby(
    ["season", "event"]
)[
    [
        "kill_participation",
        "kill_share",
        "gold_share"
    ]
].count()
```

```python
champions_season_7_and_8[
    champions_season_7_and_8[
        [
            "kill_participation",
            "kill_share",
            "gold_share"
        ]
    ].isna().any(axis=1)
][
    [
        "season",
        "event",
        "champion",
        "played_games",
        "kill_participation",
        "kill_share",
        "gold_share"
    ]
]
```

##### Conclusão sobre o preenchimento parcial

- Nas temporadas 7 e 8, as métricas `kill_participation` e `kill_share` apresentam apenas duas ausências pontuais.
- Na temporada 7, os valores ausentes pertencem ao registro da campeã Jinx no evento principal.
- Na temporada 8, os valores ausentes pertencem ao registro da campeã Nidalee no Play-In.
- Em ambos os casos, o campeão foi utilizado em apenas uma partida, mas a métrica `gold_share` permanece preenchida.
- Portanto, as ausências não parecem estar relacionadas às diferentes etapas do campeonato, já que o restante dos registros dos mesmos eventos estão preenchidos.
- Esses dois casos podem representar falhas pontuais de registro ou indisponibilidade das métricas na fonte original.

##### Síntese da qualidade dos dados de campeões

O conjunto não possui linhas completamente duplicadas. Os valores ausentes apresentam diferentes significados:

- Em `banned_games`, as ausências representam zero banimentos.
- Nos registros em que `played_games` é igual a zero, as estatísticas de desempenho não se aplicam porque o campeão não foi utilizado.
- Algumas métricas não eram registradas nas temporadas iniciais: `kill_participation`, `kill_share` e `gold_share` começam a ser disponibilizadas na temporada 4, enquanto `damage` e `damage/min` começam na temporada 10.
- Foram identificadas duas ausências pontuais em `kill_participation` e `kill_share`: Jinx na temporada 7 e Nidalee na temporada 8. A causa desses casos não pode ser determinada apenas com os dados disponíveis.

### 4.3 Qualidade dos dados de partidas
